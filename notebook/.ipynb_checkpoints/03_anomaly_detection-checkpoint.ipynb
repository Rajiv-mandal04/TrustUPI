```python
import pandas as pd
import numpy as np

# Feature engineered dataset load karo
df = pd.read_csv(r"D:\Machine Learning\upi_fraud_detection\data\processed\feature_engineered_transactions.csv")

print("Shape:", df.shape)
print(df.head())
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nColumns:")
print(df.columns.tolist())
```

```python
print(df[[
    "transaction_id",
    "sender_id",
    "amount",
    "fraud_label"
]].head())
```

```python
# Step 17.2 — IQR Calculation

Q1 = df["amount"].quantile(0.25)
Q3 = df["amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# Step 17.3 — IQR Anomaly Flag

df["amount_iqr_flag"] = (
    (df["amount"] < lower_bound) |
    (df["amount"] > upper_bound)
).astype(int)

print(df["amount_iqr_flag"].value_counts())
```

```python
print(
    "IQR Anomaly Percentage:",
    df["amount_iqr_flag"].mean() * 100
)
```

```python
# Step 17.4 — Inspected Anomalous transactions
iqr_anomalies = df[df["amount_iqr_flag"] == 1]

print("Total IQR anomalies:", len(iqr_anomalies))

print(
    iqr_anomalies[
        [
            "transaction_id",
            "sender_id",
            "amount",
            "fraud_label",
            "fraud_scenario"
        ]
    ]
    .sort_values("amount", ascending=False)
    .head(20)
)
```

```python
# Step 17.5 — Fraud vs non-fraud comparison
print(
    pd.crosstab(
        df["amount_iqr_flag"],
        df["fraud_label"],
        margins=True
    )
)
```

```python
# Step 18.1 — Features for Isolation Forest

isolation_features = [
    "amount",
    "amount_ratio",
    "amount_vs_user_avg",
    "historical_amount_zscore",
    "transactions_last_10m",
    "transactions_last_1h",
    "transactions_last_24h",
    "time_since_previous_transaction",
    "is_new_device",
    "is_new_recipient",
    "distance_from_home_km",
    "distance_from_previous_location_km",
    "travel_speed_kmph",
    "impossible_travel_flag",
    "daily_spending_ratio",
    "daily_spending_deviation",
    "high_spending_deviation_flag",
    "extreme_amount_flag"
]

X = df[isolation_features].copy()

print("Feature Shape:", X.shape)
print(X.head())
```

```python
print(X.isnull().sum()[X.isnull().sum() > 0])
```

```python
# Step 18.2 — Isolation Forest
from sklearn.ensemble import IsolationForest

model = IsolationForest(
    n_estimators=200,
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

model.fit(X)

print("Isolation Forest trained successfully.")
```

```python
# Step 18.3 — Anomaly prediction
df["isolation_prediction"] = model.predict(X)

df["isolation_anomaly_flag"] = (
    df["isolation_prediction"] == -1
).astype(int)

print(df["isolation_anomaly_flag"].value_counts())
```

```python
# Step 18.4 — Isolation Forest Evaluation
print(
    pd.crosstab(
        df["isolation_anomaly_flag"],
        df["fraud_label"],
        margins=True
    )
)
```

```python
# Step 18.5 — Anomaly percentage
print(
    "Isolation Forest Anomaly Percentage:",
    df["isolation_anomaly_flag"].mean() * 100
)
```

```python
# Step 18.6 — Top detected anomalies
isolation_anomalies = df[
    df["isolation_anomaly_flag"] == 1
]

print(
    isolation_anomalies[
        [
            "transaction_id",
            "sender_id",
            "amount",
            "fraud_label",
            "fraud_scenario"
        ]
    ]
    .sort_values("amount", ascending=False)
    .head(20)
)
```

```python
#  Step 19 — Combine IQR + Isolation Forest
df["combined_anomaly_flag"] = (
    (df["amount_iqr_flag"] == 1) |
    (df["isolation_anomaly_flag"] == 1)
).astype(int)

print(
    pd.crosstab(
        df["combined_anomaly_flag"],
        df["fraud_label"],
        margins=True
    )
)

print(
    "Combined Anomaly Percentage:",
    df["combined_anomaly_flag"].mean() * 100
)
```

```python
combined_anomalies = df[
    df["combined_anomaly_flag"] == 1
]

print("Total combined anomalies:", len(combined_anomalies))

print(
    combined_anomalies[
        [
            "transaction_id",
            "sender_id",
            "amount",
            "fraud_label",
            "fraud_scenario",
            "amount_iqr_flag",
            "isolation_anomaly_flag"
        ]
    ]
    .sort_values("amount", ascending=False)
    .head(20)
)
```

```python
# Step 20 — Combined Anomaly Score
df["anomaly_score"] = (
    df["amount_iqr_flag"] * 30
    + df["isolation_anomaly_flag"] * 40
    + df["extreme_amount_flag"] * 15
    + df["impossible_travel_flag"] * 15
)

print(df["anomaly_score"].value_counts().sort_index())
```

```python
print(
    df[
        [
            "transaction_id",
            "amount",
            "fraud_label",
            "amount_iqr_flag",
            "isolation_anomaly_flag",
            "extreme_amount_flag",
            "impossible_travel_flag",
            "anomaly_score"
        ]
    ]
    .sort_values("anomaly_score", ascending=False)
    .head(20)
)
```

```python
output_columns = [
    "transaction_id",
    "sender_id",
    "receiver_id",
    "amount",
    "timestamp",
    "fraud_label",
    "fraud_scenario",

    # Feature signals
    "amount_ratio",
    "amount_vs_user_avg",
    "historical_amount_zscore",
    "transactions_last_10m",
    "transactions_last_1h",
    "transactions_last_24h",
    "is_new_device",
    "is_new_recipient",
    "distance_from_home_km",
    "distance_from_previous_location_km",
    "travel_speed_kmph",
    "impossible_travel_flag",
    "daily_spending_ratio",
    "daily_spending_deviation",

    # Anomaly signals
    "amount_iqr_flag",
    "isolation_anomaly_flag",
    "extreme_amount_flag",
    "anomaly_score"
]

df[output_columns].to_csv(
    r"D:\Machine Learning\upi_fraud_detection\data\processed\anomaly_detection_output.csv",
    index=False
)

print("Phase 3 output saved successfully.")
print("Shape:", df[output_columns].shape)
```
